```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
```

```python
df = pd.read_csv("wearable_health_devices_performance_upto_26june2025.csv",
                 sep = ";")
df
```

```python
df = np.array(df[['User_Satisfaction_Rating', 'Performance_Score']])
df
```

```python
plt.title("Dendogram Metode Single Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'single'))
```

```python
plt.title("Dendogram Metode Average Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'average'))
```

```python
plt.title("Dendogram Metode Complete Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'complete'))
```

```python
plt.title("Dendogram Metode Ward Linkage")
dendrogram = sch.dendrogram(sch.linkage(df, method = 'ward'))
```

```python
from sklearn.cluster import AgglomerativeClustering as acg
```

```python
# 1. Metode Single Linkage
ac = acg(n_clusters = 4, linkage = 'single')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot Single Linkage')
plt.xlabel('Kepuasan Pelanggan')
plt.ylabel('Skor Performa')
plt.show()

# 2. Metode Average Linkage
ac = acg(n_clusters = 4, linkage = 'average')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot Average Linkage')
plt.xlabel('Kepuasan Pelanggan')
plt.ylabel('Skor Performa')
plt.show()

# 3. Metode Complete Linkage
ac = acg(n_clusters = 4, linkage = 'complete')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot Complete Linkage')
plt.xlabel('Kepuasan Pelanggan')
plt.ylabel('Skor Performa')
plt.show()

# 4. Metode Ward Linkage
ac = acg(n_clusters = 4, linkage = 'ward')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot Ward Linkage')
plt.xlabel('Kepuasan Pelanggan')
plt.ylabel('Skor Performa')
plt.show()
```
